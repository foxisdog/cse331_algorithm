```python
import numpy as np
import numba
import pandas as pd
import math
import networkx as nx
import random
```

```python
a = pd.read_csv("./datasets/a280.csv")
points = np.array(a[['x','y']])
inputsize = len(points)
```

```python
# # distance matrix 만들기
diff = points[:, np.newaxis, :] - points[np.newaxis, :, :]
distance_matrix = np.sqrt(np.sum(diff ** 2, axis=-1))
distance_matrix += 1
```

```python
# minimum weight perfect matching Blossom 알고리즘
graph = nx.from_numpy_array(distance_matrix)
```

```python
k = 10
centroid = np.array( [ (random.randint(a['x'].min(), a['x'].max()), random.randint(a['y'].min(), a['y'].max())) for _ in range(k)] )
```

```python
a['x'].max()
a['x'].min()
a['y'].max()
a['y'].min()
```

```python
centroid
```

```python
dist_matrix = np.linalg.norm( points[:, np.newaxis, :] - centroid[np.newaxis, :, :], axis=2)
```

```python
print(points[1])
print(centroid[0])
dist_matrix[0][1] # centroid 0 에서 points 1 까지 거리
```

```python
np.argmin(dist_matrix[220])
```

```python
label = np.full( inputsize, 0 )
```

```python
for x in range(inputsize):
    label[x] = np.argmin(dist_matrix[x])
```

```python
label
```

```python
points[1] + points[2]
```

```python
unique_labels = np.unique(label)
centroids = np.zeros((len(unique_labels), 2))

for idx, ul in enumerate(unique_labels):
    cluster_points = points[label == ul]  # 해당 레이블의 모든 점 선택
    centroids[idx] = np.mean(cluster_points, axis=0)
```
